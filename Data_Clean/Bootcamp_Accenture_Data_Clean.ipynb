```python
import sqlite3

# criando ou conectando ao banco

conexao = sqlite3.connect("dados.db")
cursor = conexao.cursor()

cursor.execute("""
  CREATE TABLE IF NOT EXISTS usuarios_vip (
    nome TEXT,
    idade INTEGER
    )
  """)

conexao.commit()
cursor.close()
conexao.close()


```

```python
# Inserindo os dados no banco

import sqlite3

conexao = sqlite3.connect("dados.db")

cursor = conexao.cursor()

cursor.execute(
    "INSERT INTO usuarios_vip VALUES (?, ?)",
    ("Lunardi", 37)
)

conexao.commit()
cursor.close()
conexao.close()
```

```python
# Leitura com fetchall()
# Lendo todos os resultados de uma vez
import sqlite3

conexao = sqlite3.connect("dados.db")

cursor = conexao.cursor()

cursor.execute("Select * FROM usuarios_vip")

# retorna uma lista com todas as linhas(em Tuplas)
dados = cursor.fetchall()

print(dados)

cursor.close()
conexao.close()
```

```python
# Leitura iterativa no cursor
# Lendo linha por linha diretamente do cursor
import sqlite3

conexao = sqlite3.connect("dados.db")
cursor = conexao.cursor()

cursor.execute("SELECT * FROM usuarios_vip")
for linha in cursor:
    print(linha)

cursor.close()
conexao.close()
```

```python
import sqlite3

# Dropando tabela do banco

conexao = sqlite3.connect("dados.db")
cursor = conexao.cursor()

cursor.execute("""
  DROP TABLE IF EXISTS usuarios_vip
  """)

conexao.commit()
cursor.close()
conexao.close()
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
# Substitui os valores nulos da coluna "Age" pela média da própria coluna,
# garantindo que não haja mais dados faltantes nessa variável.
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Verifica novamente a quantidade de valores nulos em "Age"
# para confirmar que todos foram preenchidos.
df["Age"].isnull().sum()
```

```python
df["Age_log"] = df["Age"].apply(lambda x: x)

print(df["Age_log"])
```

```python
# Escala de dados
# É utilizado quando precisamos colocar os dados na mesma escala

df["Fare_normalized"] = (df["Fare"] - df["Fare"].min()) / (df["Fare"].max() - df["Fare"].min())

df.head()
```

```python
def limpar_dados(df):
  df["Age"].fillna(df["Age"].mean(), inplace=True)

  df["Fare_normalized"] = (
    df["Fare"] - df["Fare"].min()
  ) / (df["Fare"].max() - df["Fare"].min())

  return df
```

```python
df = limpar_dados(df)

df.head()
```

```python
# Documentação e Validação

def validar_dataset(df):

  print("Linhas:", df.shape[0])
  print("Colunas:", df.shape[1])
  print("\nValores Ausentes:")

  print(df.isnull().sum())
```

```python
validar_dataset(df)
```

```python
# Tratamento simplista da coluna "Cabin":
# Substitui valores ausentes (NaN) por "Unknown",
# garantindo que não haja linhas com NaN e mantendo a coluna original.
df["Cabin"] = df["Cabin"].fillna("Unknown")

# Conferir novamente se ainda existem valores ausentes
print(df.isnull().sum())
```
